# Matrix decomposition exercises

`````{exercise}
:label: ex-lu

Solve the following system of linear equations using {prf:ref}`LU decomposition<lu-theorem>`.

$$\begin{align*}
    2 x_1 + 3 x_2 -   x_3   &=  4,\\
    4 x_1 + 9 x_2 -   x_3   &= 18,\\
            3 x_2 + 2 x_3 &= 11.
\end{align*} $$

```{dropdown} Solution

The coefficient matrix is 

$$ A = \begin{pmatrix} 2 & 3 & -1 \\ 4 & 9 & -1 \\ 0 & 3 & 2 \end{pmatrix}. $$

Compute the LU decomposition of $A$

$$ \begin{align*}
  j &= 1: & u_{11} &= a_{11} = 2, \\
  && \ell_{21} &= \frac{1}{u_{11}}a_{21} = \frac{1}{2}(4) = 2, \\
  && \ell_{31} &= \frac{1}{u_{11}}a_{31} = \frac{1}{2}(0) = 0, \\
  j &= 2: & u_{12} &= a_{12} = 3, \\
  && u_{22} &= a_{22} - \ell_{21}u_{12} = 9 - 2(3) = 3, \\
  && \ell_{32} &= \frac{1}{u_{22}}(a_{32} - \ell_{31}u_{12}) = \frac{1}{3}(3 - 0(3)) = 1, \\
  j &= 3: & u_{13} &= a_{13} = -1, \\
  && u_{23} &= a_{23} - \ell_{21}u_{13} = -1 - 2(-1) = 1, \\
  && u_{33} &= a_{33} - \ell_{31}u_{13} - \ell_{32}u_{23} = 2 - 0(-1) - 1(1) = 1,
\end{align*} $$

therefore

$$ \begin{align*}
  L &= \begin{pmatrix}
    1 & 0 & 0 \\
    2 & 1 & 0 \\
    0 & 1 & 1
  \end{pmatrix}, &
  U &= \begin{pmatrix}
    2 & 3 & -1 \\
    0 & 3 & 1 \\
    0 & 0 & 1
  \end{pmatrix}.
\end{align*} $$

Solving $L \mathbf{y} = \mathbf{b}$ using forward substitution

$$ \begin{align*}
  \begin{pmatrix}
      1 & 0 & 0 \\
      2 & 1 & 0 \\
      0 & 1 & 1
    \end{pmatrix}
    \begin{pmatrix} y_1 \\ y_2 \\ y_3 \end{pmatrix}
    =
    \begin{pmatrix} 4 \\ 18 \\ 11 \end{pmatrix} 
    \quad \implies \quad
    \begin{aligned} 
      y_1 &= 4, \\
      y_2 &= 18 - 2(4) = 10, \\
      y_3 &= 11 - 10 = 1.
    \end{aligned}
\end{align*} $$

Solving $U \mathbf{x} = \mathbf{y}$ using back substitution

$$ \begin{align*}
  \begin{pmatrix}
    2 & 3 & -1 \\
    0 & 3 & 1 \\
    0 & 0 & 1
  \end{pmatrix}
  \begin{pmatrix} x_1 \\ x_2 \\ x_3 \end{pmatrix}
  =
  \begin{pmatrix} 4 \\ 10 \\ 1 \end{pmatrix}
  \quad \implies \quad
  \begin{aligned}
    x_3 &= 1, \\
    x_2 &= \frac{1}{3}(10 - 1) = 3, \\
    x_1 &= \frac{1}{2}(4 - 3(3) + 1) = -2.
  \end{aligned}
\end{align*} $$
```
`````


``````python
import numpy as np

def lu(A):
    n = A.shape[0]
    L, U = np.eye(n), np.zeros((n, n))
    for j in range(n):
        for i in range(n):
            sum_ = 0
            if i <= j:
                for k in range(i):
                    sum_ += L[i,k] * U[k,j]
        
                U[i,j] = A[i,j] - sum_   
            
            else:         
                for k in range(j):
                    sum_ += L[i,k] * U[k,j]
                    
                L[i,j] = (A[i,j] - sum_) / U[j,j]
    
    return L, U


A = np.array([[2, 3, -1], [4, 9, -1], [0, 3, 2]])
b = np.array([4, 18, 11])
L, U = lu(A)

print('(a)')
print(f'L = {L}')
print(f'\nU = {U}')
print(f'\ny = {np.linalg.solve(L, b)}')
print(f'\nx = {np.linalg.solve(A, b)}')
``````

````{exercise}
:label: ex-lup

Solve the following system of linear equations using [LU decomposition with partial pivoting](lup-section).

$$ \begin{align*}
    x_1 + 3x_2 + 2x_3 &= 16, \\
    2x_1 + x_2 &= 0, \\
    3x_1 + x_2 - x_3 &= -5.
\end{align*} $$

```{dropdown} Solution
The coefficient matrix is

$$ A = \begin{pmatrix} 1 & 3 & 2 \\ 2 & 1 & 0 \\ 3 & 1 & -1 \end{pmatrix}.$$

Performing partial pivoting on $A$

$$ \begin{align*}
  & \begin{pmatrix} 
    1 & 3 & 2 \\ 
    2 & 1 & 0 \\ 
    3 & 1 & -1 
  \end{pmatrix}
  \begin{matrix} R_1 \leftrightarrow R_3  \\ \phantom{x} \\ \phantom{x} \end{matrix} &
  \longrightarrow
  & \begin{pmatrix} 
    3 & 1 & -1 \\
    2 & 1 & 0 \\ 
    1 & 3 & 2
  \end{pmatrix}
  \begin{matrix} \\ R_2 \leftrightarrow R_3 \\ \phantom{x} \end{matrix} \\ \\
  \longrightarrow
  & \begin{pmatrix} 
    3 & 1 & -1 \\
    1 & 3 & 2 \\
    2 & 1 & 0
  \end{pmatrix} = PA.
\end{align*} $$

So the permutation matrix is

$$ \begin{align*}
  & \begin{pmatrix} 
    1 & 0 & 0 \\ 
    0 & 1 & 0 \\ 
    0 & 0 & 1 
  \end{pmatrix}
  \begin{matrix} R_1 \leftrightarrow R_3  \\ \phantom{x} \\ \phantom{x} \end{matrix} &
  \longrightarrow
  & \begin{pmatrix} 
    0 & 0 & 1 \\ 
    0 & 1 & 0 \\ 
    1 & 0 & 0
  \end{pmatrix}
  \begin{matrix} \\ R_2 \leftrightarrow R_3 \\ \phantom{x} \end{matrix} \\ \\
  \longrightarrow
  & \begin{pmatrix} 
    0 & 0 & 1 \\ 
    1 & 0 & 0 \\ 
    0 & 1 & 0 
  \end{pmatrix} = P.
\end{align*} $$

Computing the LU decomposition of $PA$

$$ \begin{align*}
  j &= 1: & u_{11} &= a_{11} = 3, \\
  && \ell_{21} &= \frac{1}{u_{11}}a_{21} = \frac{1}{3}(1) = \frac{1}{3}, \\
  && \ell_{31} &= \frac{1}{u_{11}}a_{31} = \frac{1}{3}(2) = \frac{2}{3}, \\
  j &= 2: & u_{12} &= a_{12} = 1, \\
  && u_{22} &= a_{22} - \ell_{21}u_{12} = 3 - \frac{1}{3}(1) = \frac{8}{3}, \\
  && \ell_{32} &= \frac{1}{u_{22}}(a_{32} - \ell_{31}u_{12}) = \frac{3}{8}\left(1 - \frac{2}{3}(1)\right) = \frac{1}{8}, \\
  j &= 3: & u_{13} &= a_{13} = -1, \\
  && u_{23} &= a_{23} - \ell_{21}u_{13} = 2 - \frac{1}{3}(-1) = \frac{7}{3}, \\
  && u_{33} &= a_{33} - \ell_{31}u_{13} - \ell_{32}u_{23} = 0 - \frac{2}{3}(-1) - \frac{1}{8}\left( \frac{7}{3} \right) = \frac{3}{8},
\end{align*} $$

therefore 

$$ \begin{align*}
L &= 
  \begin{pmatrix}
    1 & 0 & 0 \\
    \frac{1}{3} & 1 & 0 \\
    \frac{2}{3} & \frac{1}{8} & 1
  \end{pmatrix}, 
  \qquad
  U &= 
  \begin{pmatrix}
    3 & 1 & -1 \\
    0 & \frac{8}{3} & \frac{7}{3} \\
    0 & 0 & \frac{3}{8}
  \end{pmatrix}, 
  \qquad
  P &= \begin{pmatrix} 
    0 & 0 & 1 \\ 
    1 & 0 & 0 \\ 
    0 & 1 & 0 
  \end{pmatrix}.
\end{align*} $$

Solving $L \mathbf{y} = P\mathbf{b}$ using forward substitution

$$ 
\begin{pmatrix}
  1 & 0 & 0 \\
  \frac{1}{3} & 1 & 0 \\
  \frac{2}{3} & \frac{1}{8} & 1
\end{pmatrix}
\begin{pmatrix} y_1 \\ y_2 \\ y_3 \end{pmatrix}
= 
\begin{pmatrix} -5 \\ 16 \\ 0 \end{pmatrix}
\quad \implies \quad
\begin{aligned}
  y_1 &= -5 \\
  y_2 &= 16 - \frac{1}{3}(-5) = \frac{53}{3}, \\
  y_3 &= 0 - \frac{2}{3}(-5) - \frac{1}{8}\left( \frac{53}{3} \right) = \frac{9}{8}.
\end{aligned}
$$

Solving $U \mathbf{x} = \mathbf{y}$ using back substitution

$$ 
\begin{pmatrix}
  3 & 1 & -1 \\
  0 & \frac{8}{3} & \frac{7}{3} \\
  0 & 0 & \frac{3}{8}
\end{pmatrix}
\begin{pmatrix} x_1 \\ x_2 \\ x_3 \end{pmatrix}
=
\begin{pmatrix} -5 \\ \frac{53}{3} \\ \frac{9}{8} \end{pmatrix}
\quad \implies \quad
\begin{aligned}
  x_3 &= \frac{8}{3}\left( \frac{9}{8} \right) = 3, \\
  x_2 &= \frac{3}{8}\left( \frac{53}{3} - \frac{7}{3}(3)\right)= 4, \\
  x_1 &= \frac{1}{3}(-5 - 4 + 3) = -2.
\end{aligned}
$$
````

``````python
def partial_pivot(A):
    n = A.shape[0]
    P = np.eye(n)
    for j in range(n):
        maxpivot, maxpivotrow = abs(A[j,j]), j
        for i in range(j + 1, n):
            if abs(A[i,j]) > maxpivot:
                maxpivot, maxpivotrow = abs(A[i,j]), i

        A[[j,maxpivotrow]] = A[[maxpivotrow,j]] 
        P[[j,maxpivotrow]] = P[[maxpivotrow,j]]

    return P


A = np.array([[0, 3, 2], [4, 9, -1], [2, 3, -1]])
b = np.array([11, 18, 4])

P = partial_pivot(A)
L, U = lu(A)
y = np.linalg.solve(L, np.dot(P, b))
x = np.linalg.solve(U, y)

print(f'P = {P}')
print(f'\nPA = {A}')
print(f'\nL = {L}')
print(f'\nU = {U}')
print(f'\ny = {y}')
print(f'\nx = {x}')
``````

````{exercise}
:label: ex-cholesky

Solve the following systems of linear equations using [Cholesky decomposition](cholesky-section).

$$ \begin{align*}
    16x_1 +16x_2 +4x_3 &=-8,\\
    16x_1 +25x_2 +10x_3 &=-47,\\
    4x_1 +10x_2 +6x_3 &=-30.
\end{align*} $$

```{dropdown} Solution

The coefficient matrix is

$$ A = \begin{pmatrix}
  16 & 16 & 4 \\
  16 & 25 & 10 \\
  4 & 10 & 6
\end{pmatrix}. $$

Checking whether $A$ is a positive definite matrix

$$ \begin{align*}
  \det(16) &= 16, \\
  \det \begin{pmatrix} 16 & 16 \\ 16 & 25 \end{pmatrix} &= 144, \\
  \det \begin{pmatrix}
    16 & 16 & 4 \\
    16 & 25 & 10 \\
    4 & 10 & 6
  \end{pmatrix}
  &= 16 \begin{vmatrix} 25 & 10 \\ 10 & 6 \end{vmatrix} 
  - 16 \begin{vmatrix} 16 & 10 \\ 4 & 6 \end{vmatrix}
  + 4 \begin{vmatrix} 16 & 25 \\ 4 & 10 \end{vmatrix} \\
  &= 16(50) - 16(56) + 4(60) = 144,
\end{align*} $$

therefore $A$ is positive definite. Computing the Cholesky decomposition of $A$

$$ \begin{align*}
  j &= 1: & \ell_{11} &= \sqrt{a_{11}} = \sqrt{16} = 4, \\
  && \ell_{21} &= \frac{1}{\ell_{11}}a_{21} = \frac{1}{4}(16) = 4, \\
  && \ell_{31} &= \frac{1}{\ell_{11}}a_{31} = \frac{1}{4}(4) = 1, \\
  j &= 2: & \ell_{22} &= \sqrt{a_{22} - \ell_{21}^2} = \sqrt{25 - 4^2} = 3, \\
  && \ell_{32} &= \frac{1}{\ell_{22}}(a_{32} - \ell_{31}\ell_{21}) = \frac{1}{3}(10 - 1(4)) = 2, \\
  j &= 3: & \ell_{33} &+ \sqrt{a_{33} - \ell_{31}^2 - \ell_{32}^2} = \sqrt{6 - 1^2 - 2^2} = 1,
\end{align*} $$

therefore 

$$ \begin{align*}
  L = 
  \begin{pmatrix}
    4 & 0 & 0 \\
    4 & 3 & 0 \\
    1 & 2 & 1 
  \end{pmatrix}.
\end{align*} $$

Solving $L \mathbf{y} = \mathbf{b}$ using forward substitution

$$
\begin{pmatrix}
  4 & 0 & 0 \\
  4 & 3 & 0 \\
  1 & 2 & 1 
\end{pmatrix}
\begin{pmatrix} y_1 \\ y_2 \\ y_3 \end{pmatrix}
=
\begin{pmatrix} -8 \\ -47 \\ -30 \end{pmatrix}
\quad \implies \quad
\begin{aligned}
  y_1 &= \frac{1}{4}(-8) = -2, \\
  y_2 &= \frac{1}{3}(-47 - 4(-2)) = - 13, \\
  y_3 &= -30 - (-2) - 2(-13) = -2.
\end{aligned}
$$

Solving $L^\mathsf{T} \mathbf{x} = \mathbf{y}$ using back substitution

$$
\begin{pmatrix}
  4 & 4 & 1 \\
  0 & 3 & 2 \\
  0 & 0 & 1
\end{pmatrix}
\begin{pmatrix} x_1 \\ x_2 \\ x_3 \end{pmatrix}
=
\begin{pmatrix} -2 \\ -13 \\ -2 \end{pmatrix}
\quad \implies \quad
\begin{aligned}
  x_3 &= -2, \\
  x_2 &= \frac{1}{3}(-13 - 2(-2)) = -3, \\
  x_1 &= \frac{1}{4}(-2 -4(3) - (-2)) = 3.
\end{aligned}
$$
```
````

``````python
def cholesky(A):
    n = A.shape[0]
    L = np.zeros((n, n))   
    for j in range(n):
        sum_ = 0
        for k in range(j):
            sum_ += L[j,k] ** 2
        
        L[j,j] = np.sqrt(A[j,j] - sum_)
                
        for i in range(j+1, n):
            sum_ = 0
            for k in range(j):
                sum_ += L[i,k] * L[j,k]
            
            L[i,j] = (A[i,j] - sum_) / L[j,j]
    
    return L


A = np.array([[16, 16, 4], [16, 25, 10], [4, 10, 6]])
b = np.array([-8, -47, -30])
L = cholesky(A)
y = np.linalg.solve(L, b)
x = np.linalg.solve(L.T, y)

print(f'L = {L}')

print(f'\ny = {y}')
print(f'\nx = {x}')
``````
