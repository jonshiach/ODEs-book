(b) The exact solution is $y(1) = 0.554504$ and numerical solutions are $y(1, h= 0.2) = 0.554686$ and $y(1,h = 0.1) = 0.554514$ so the GTE are

$$ \begin{align*}
  E(h = 0.2) = |0.554504 - 0.554686| = 1.82 \times 10^{-4}$, \\
  E(h = 0.1) = |0.554504 - 0.554515| = 9.60 \times 10^{-5}$,
\end{align*} $$

therefore

$$ \text{order} \approx \frac{\log \left( \dfrac{E(h1)}{E(h2)}\right)}{\log \left( \dfrac{h_1}{h_2}\right)} = \frac{\log \left( \dfrac{1.82 \times 10^{-4}}{9.60 \times 10^{-5}} \right)}{\log (\left( \dfrac{0.2}{0.1} \right))} = 4.24. $$

```````python
# (a) 
import numpy as np

def f(t, y): 
    return t - 2 * y

h = 0.2
nsteps = 5

t = np.arange(0, 1 + h, h)
y = np.zeros(t.shape)
y[0] = 2

for n in range(nsteps):
    k1 = f(t[n], y[n])
    k2 = f(t[n] + 1/4 * h, y[n] + 1/4 * h * k1)
    k3 = f(t[n] + 1/2 * h, y[n] + 1/2 * h * k2)
    k4 = f(t[n] + h, y[n] + h * (k1 - 2 * k2 + 2 * k3))

    y[n+1] = y[n] + h / 6 * (k1 + 4 * k3 + k4)

print(t)
print(y)
```````

```````python
# (b) 
h = 0.1
nsteps = 10

t = np.arange(0, 1 + h, h)
y2 = np.zeros(t.shape)
y2[0] = 2

for n in range(nsteps):
    k1 = f(t[n], y2[n])
    k2 = f(t[n] + 1/4 * h, y2[n] + 1/4 * h * k1)
    k3 = f(t[n] + 1/2 * h, y2[n] + 1/2 * h * k2)
    k4 = f(t[n] + h, y2[n] + h * (k1 - 2 * k2 + 2 * k3))

    y2[n+1] = y2[n] + h / 6 * (k1 + 4 * k3 + k4)


def exact(t):
    return t / 2 - 1/4 + 9/4 * np.exp(-2 * t)

E1 = abs(exact(1) - y[-1])
E2 = abs(exact(1) - y2[-1])

print(f"y(1) = {exact(1):0.6f}")
print(f"y(1, h = 0.2) = {y[-1]:0.6f}")
print(f"y(1, h = 0.2) = {y2[-1]:0.6}")
print(f"GTE (h = 0.2) = {E1:0.2e}")
print(f"GTE (h = 0.1) = {E2:0.2e}")

order = np.log(E1 / E2) / np.log(0.2 / 0.1)

print(f"order estimate = {order}")
```````
