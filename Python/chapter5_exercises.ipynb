#### Exercise 5.1

(a) &emsp; $p(t) = -\frac{4}{t}$, $q(t) = \frac{2}{t^2}$ and $r(t) = \frac{2\ln(t)}{t^3}$. $p(t)$, $q(t)$ and $r(t)$ are all continuous in the range $[1, 2]$ and $q(t) > 0$ for all $t$ so this BVP has a unique solution.

(b) &emsp; $p(t) = -(t + 1)$, $q(t) = \cos(t)$ and $r(t) = e^t$. $p(t)$, $r(t)$ and $q(t)$ are all continuous in the range $[0, 2]$ but $q(t) < 0$ in the range $[\frac{\pi}{2}, 2]$ so this BVP does not have a unique solution.

(c) &emsp; $p(t) = 0$, $q(t) = (t^3 + 5)$ and $r(t) = \sin(t)$. $p(t)$, $r(t)$ and $q(t)$ are all continuous in the range $[0, 1]$ and $q(t) > 0$ in the same range so this BVP has a unique solution.

(d) &emsp; $p(t) = 0, $q(t) = 5e^t$ and $r(t) = e^t\sin(3t)$. $p(t)$, $q(t)$ and $r(t)$ are all continuous in the range $[0, 1]$ and $q(t) > 0$ for all $t$ so this BVP has a unique solution.

#### Exercise 5.2

```python
import numpy as np
import matplotlib.pyplot as plt

def euler(f, tspan, y0, h):
    
    N = len(y0)
    nsteps = int((tspan[1] - tspan[0]) / h)
    t = np.zeros(nsteps + 1)
    y = np.zeros((N, nsteps + 1))
        
    t[0] = tspan[0]
    y[:,0] = y0
    
    for n in range(nsteps):
        y[:,n+1] = y[:,n] + h * f(t[n], y[:,n])
        t[n+1] = t[n] + h

    return t, y.T


# Define ODE function
def f(t, y):
    return np.array([y[1], -4*y[0]])


# Define BVP parameters
tspan = [0, 1]  # boundaries of the t domain
bvals = [0, 1]  # boundary values
h = 0.2         # step length

# Calculate the solution to the IVP using s = 1 and s = 2
t, y1 = euler(f, tspan, [bvals[0], 1], h)
t, y2 = euler(f, tspan, [bvals[0], 2], h)

# Print tables of solutions
print("|   t   |     y1     |     y2    |")
print("|:-----:|:----------:|:----------:|")
for n in range(len(t)):
    print(f"| {t[n]:5.2f} | {y1[n,0]:10.6f} | {y1[n,1]:10.6f} |")

print("\n|   t   |     y1     |     y2     |")
print("|:-----:|:----------:|:----------:|")
for n in range(len(t)):
    print(f"| {t[n]:5.2f} | {y2[n,0]:10.6f} | {y2[n,1]:10.6f} |")
```

#### Exercise 5.4

$$y'' + 4y = 0, \qquad y(0) = 0, \qquad y(1) = 1.$$

Using $y'' = \dfrac{y_{i-1} - 2y_i + y_{i+1}}{h^2}$ then

$$ \begin{align*}
    \dfrac{y_{i-1} - 2y_i + y_{i+1}}{h^2} + 4y_i &= 0 \\
    y_{i-1} + (4h^2 - 2)y_i + y_{i+1} &= 0.
\end{align*}$$

Since $h = 0.2$, $y_0 = 0$ and $y_5 = 1$ then

$$ \begin{align*}
    y_0 &= 0, \\
    y_0 - 1.84 y_1 + y_2 &= 0, \\
    y_1 - 1.84 y_2 + y_3 &= 0, \\
    y_2 - 1.84 y_3 + y_4 &= 0, \\
    y_3 - 1.84 y_4 + y_5 &= 0, \\
    y_5 &= 1,
\end{align*} $$

which can be written as the matrix equation

$$ \begin{align*}
    \begin{pmatrix} 
        1 & 0 & 0 & 0 & 0 & 0 \\
        1 & -1.84 & 1 & 0 & 0 & 0 \\
        0 & 1 & -1.84 & 1 & 0 & 0 \\
        0 & 0 & 1 & -1.84 & 0 & 0 \\
        0 & 0 & 0 & 1 & -1.84 & 1 \\
        0 & 0 & 0 & 0 & 0 & 1
    \end{pmatrix} 
    \begin{pmatrix} y_0 \\ y_1 \\ y_2 \\ y_3 \\ y_4 \\ y_5 \end{pmatrix}
    =
    \begin{pmatrix} 0 \\ 0 \\ 0 \\ 0 \\ 0 \\ 1 \end{pmatrix}.
\end{align*} $$

Solving using Python:

```python
# Define linear system
n = int((tspan[1] - tspan[0]) / h) + 1
A = np.eye(n)
b = np.zeros(n)
b[0], b[-1] = bvals[0], bvals[1]

for i in range(1, n - 1):
  A[i,i-1] = 1
  A[i,i] = 4 * h ** 2 - 2
  A[i,i+1] = 1

# Solve linear system
y = np.linalg.solve(A, b)
t = np.arange(tspan[0], tspan[1] + h, h)

# Print tables of solution
print("|   t   |     y      |")
print("|:-----:|:----------:|")
for n in range(len(t)):
    print(f"| {t[n]:5.2f} | {y[n]:10.6f} |")
```

```python
# Define exact solution
def exact(t):
    return np.sin(2 * t) / np.sin(2)

# Calculate exact solution
texact = np.linspace(tspan[0], tspan[1], 100)
yexact = exact(texact)

# Plot solutions
fig, ax = plt.subplots(figsize=(6,4))
plt.plot(texact, yexact, 'k', label="Exact")
plt.plot(t, y3[:,0], 'b-o', label="Euler")
plt.plot(t, y, 'r-o', label='Finite-difference')
plt.xlabel('$t$')
plt.ylabel('$y(t)$')
plt.legend()
plt.show()
```
