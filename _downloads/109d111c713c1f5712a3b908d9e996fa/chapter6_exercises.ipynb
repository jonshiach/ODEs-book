#### Exercise 6.1

```python
import sympy as sp

A = sp.Matrix([[2, 3, -1], [4, 9, -1], [0, 3, 2]])
b = sp.Matrix([4, 18, 11])

L, U = sp.eye(3), sp.zeros(3)

# j = 1
U[0,0] = A[0,0]
L[1,0] = A[1,0] / U[0,0]
L[2,0] = A[2,0] / U[0,0]

# j = 2
U[0,1] = A[0,1]
U[1,1] = A[1,1] - L[1,0] * U[0,1]
L[2,1] = (A[2,1] - L[2,0] * U[0,1]) / U[1,1]

# j = 3
U[0,2] = A[0,2]
U[1,2] = A[1,2] - L[1,0] * U[0,2]
U[2,2] = A[2,2] - L[2,0] * U[0,2] - L[2,1] * U[1,2]

print(f"L =")
sp.pprint(L)
print(f"\nU =")
sp.pprint(U)
```

Solve $L\mathbf{y} = \mathbf{b}$.

```python
y = sp.zeros(3, 1)
y[0] = b[0]
y[1] = b[1] - L[1,0] * y[0]
y[2] = b[2] - L[2,0] * y[0] - L[2,1] * y[1]

print(f"\ny =")
sp.pprint(y)
```

Solve $U \mathbf{x} = \mathbf{y}$.

```python
# Solve Ux = y
x = sp.zeros(3, 1)
x[2] = y[2] / U[2,2]
x[1] = (y[1] - U[1,2] * x[2]) / U[1,1]
x[0] = (y[0] - U[0,1] * x[1] - U[0,2] * x[2]) / U[0,0]

print(f"\nx =")
sp.pprint(x)
```

Calculate Cholesky decomposition of $A$.

```python
L = sp.zeros(3, 3)
L[0,0] = sp.sqrt(A[0,0])
L[1,0] = A[1,0] / L[0,0]
L[2,0] = A[2,0] / L[0,0]
L[1,1] = sp.sqrt(A[1,1] - L[1,0]**2)
L[2,1] = (A[2,1] - L[2,0] * L[1,0]) / L[1,1]
L[2,2] = sp.sqrt(A[2,2] - L[2,0]**2 - L[2,1]**2)

print("L =")
sp.pprint(L)
```

Solve $L\mathbf{y} = \mathbf{b}$.

```python
y = sp.zeros(3, 1)
y[0] = b[0] / L[0,0]
y[1] = (b[1] - L[1,0] * y[0]) / L[1,1]
y[2] = (b[2] - L[2,0] * y[0] - L[2,1] * y[1]) / L[2,2]

print("y =")
sp.pprint(y)
```

Solve $L^\mathsf{T} \mathbf{x} = \mathbf{y}$.

```python
x = sp.zeros(3, 1)
L = L.T
x[2] = y[2] / L[2,2]
x[1] = (y[1] - L[1,2] * x[2]) / L[1,1]
x[0] = (y[0] - L[0,1] * x[1] - L[0,2] * x[2]) / L[0,0]

print("x =")
sp.pprint(x)
```

#### Exercise 6.4

Exercise 6.1 solution

```python
import numpy as np

def lu(A):
    n = A.shape[0]
    L, U = np.eye(n), np.zeros((n, n))
    
    for j in range(n):
        for i in range(j+1):
            sum_ = 0
            for k in range(i):
                sum_ += L[i,k] * U[k,j]
            
            U[i,j] = A[i,j] - sum_
        
        for i in range(j+1, n):
            sum_ = 0
            for k in range(i):
                sum_ += L[i,k] * U[k,j]

            L[i,j] = (A[i,j] - sum_) / U[j,j]
                
    return L, U


def forward_substitution(L, b):
    n = L.shape[0]
    x = np.zeros(n)
    for i in range(n):
        sum_ = 0
        for j in range(i):
            sum_ += L[i,j] * x[j]
            
        x[i] = (b[i] - sum_) / L[i,i]
    
    return x


def back_substitution(U, b):
    n = U.shape[0]
    x = np.zeros(n)
    x[-1] = b[-1] / U[-1,-1]
    for i in range(n - 2, -1, -1):
        sum_ = 0
        for j in range(i + 1, n):
            sum_ += U[i,j] * x[j]
            
        x[i] = (b[i] - sum_) / U[i,i]
        
    return x


A = np.array([[ 2, 3, -1] , [4, 9, -1 ], [ 0, 3, 2 ]])
b = np.array([ 4, 18, 11 ])

L, U = lu(A)
y = forward_substitution(L, b)
x = back_substitution(U, y)

print(f'L = \n{L}')
print(f'\nU = \n{U}')
print(f'\ny = {y}')
print(f'\nx = {x}')
```

Exercise 6.2 solution

```python
def partial_pivot(A):
    n = A.shape[0]
    P = np.eye(n)
    for j in range(n):
        maxpivot, maxpivotrow = abs(A[j,j]), j
        for i in range(j + 1, n):
            if abs(A[i,j]) > maxpivot:
                maxpivot, maxpivotrow = abs(A[i,j]), i

        A[[j,maxpivotrow]] = A[[maxpivotrow,j]] 
        P[[j,maxpivotrow]] = P[[maxpivotrow,j]]

    return P


A = np.array([[ 1, 3, 2 ], [ 2, 1, 0 ], [3, 1, -1 ]])
b = np.array([ 16, 0, -5 ])

P = partial_pivot(A)
L, U = lu(A)
y = forward_substitution(L, np.dot(P, b))
x = back_substitution(U, y)

print(f'L = \n{L}')
print(f'\nU = \n{U}')
print(f'\ny = {y}')
print(f'\nx = {x}')
```

Exercise 6.3 solution

```python
def cholesky(A):
    n = A.shape[0]
    L = np.zeros((n, n))   
    for j in range(n):
        sum_ = 0
        for k in range(j):
            sum_ += L[j,k] ** 2
        
        L[j,j] = np.sqrt(A[j,j] - sum_)
                
        for i in range(j+1, n):
            sum_ = 0
            for k in range(j):
                sum_ += L[i,k] * L[j,k]
            
            L[i,j] = (A[i,j] - sum_) / L[j,j]
    
    return L


A = np.array([[ 16, 16, 4 ], [ 16, 25, 10 ], [ 4, 10, 6 ]])
b = np.array([ -8, -47, -30 ])

L = cholesky(A)
y = forward_substitution(L, b)
x = back_substitution(L.T, y)

print(f'L = \n{L}')
print(f'\ny = {y}')
print(f'\nx = {x}')
```

```python
A = sp.Matrix([[3, 0], [4, 5]])

Q, R = sp.eye(2), A.copy()

# j = 1
x = R[:,0]
v = x + x.norm() * sp.Matrix([[1], [0]])
H = sp.eye(2) - 2 * (v * v.T) / (v.T * v)[0]
R = H * R
Q = Q * H

print("Q =")
sp.pprint(Q)
print("\nR =")
sp.pprint(R)
```

The first diagonal element of $R$ is negative, so multiply the first row of $R$ by $-1$ and the first column of $Q$ by $-1$.

```python
R[0,:] *= -1
Q[:,0] *= -1

print("Q =")
sp.pprint(Q)
print("\nR =")
sp.pprint(R)
```

First columns of $R$ is already parallel to $e_1$, so we move to the second column.

```python
x = R[1:,1]
v = x + x.norm() * sp.Matrix([[1], [0]])
Hdash = sp.eye(2) - 2 * (v * v.T) / (v.T * v)[0]
H = sp.eye(3)
H[1:,1:] = Hdash
R = H * R
Q = Q * H

print("Q =")
sp.pprint(Q)
print("\nR =")
sp.pprint(R)
```

Second and third main diagonal elements of $R$ are negative, so multiply the second and third rows of $R$ by $-1$ and the second and third columns of $Q$ by $-1$.

```python
R[1:,:] *= -1
Q[:,1:] *= -1

print("Q =")
sp.pprint(Q)
print("\nR =")
sp.pprint(R)
```

```python
def qr_gramschmidt(A):
  
    n = A.shape[1]
    Q, R = np.copy(A).astype(float), np.zeros((n,n))

    for j in range(n):
        for i in range(j):
            R[i,j] = np.dot(Q[:,i], A[:,j])
            Q[:,j] = Q[:,j] - R[i,j] * Q[:,i]
        
        R[j,j] = np.linalg.norm(Q[:,j])
        Q[:,j] = Q[:,j] / R[j,j]

    for i in range(n):
        if R[i,i] < 0:
            R[i,:] *= -1
            Q[:,i] *= -1

    return Q, R
    

A = np.array([[ 1, 1 ], [ -1, 0 ] ])

Q, R = qr_gramschmidt(A)

print(f'Q = \n{Q}')
print(f'\nR = \n{R}')
```

```python
def qr_householder(A):
    
    m, n = A.shape
    Q, R = np.eye(m), np.copy(A).astype(float)

    for i in range(m - 1):
        x = R[i:,i]
        v = np.array([x + np.sign(x[0]) *  np.linalg.norm(x) * np.eye(m - i)[:,0]]).T
        H = np.eye(m)
        H[i:,i:] -= 2 * np.dot(v, v.T) / np.dot(v.T, v)
        R = np.dot(H, R)
        Q = np.dot(Q, H)

    for i in range(n):
        if R[i,i] < 0:
            R[i,:] *= -1
            Q[:,i] *= -1

    return Q, R


A = np.array([[ 3, 0 ], [ 4, 5 ]])

Q, R = qr_householder(A)

print(f'Q = \n{Q}')
print(f'\nR = \n{R}')
```
