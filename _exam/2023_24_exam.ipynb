```python
import numpy as np

def f(t, y):
    return -3 * t * y

h =  0.1
t = 0
y = 1

nsteps = int(1/2 / h)

print("  t   |    y \n-----------------")

for n in range(nsteps):
    k1 = f(t, y)
    k2 = f(t + 1/3 * h, y + 1/3 * h * k1)
    k3 = f(t + 2/3 * h, y + h * (-1/3 * k1 + k2))
    k4 = f(t + h, y + h * (k1 - k2 + k3))
    y = y + h / 8 * (k1 + 3 * k2 + 3 * k3 + k4)
    t = t + h

    print(f" {t:0.2f} | {y:0.6f} ")
```

```python
A = np.array([[1, 3, 2], [2, 5, 1], [3, 7, 2]])

L = np.tril(A, -1)
U = np.triu(A, 1)
D = A - L - U

TJ = -np.dot(np.linalg.inv(D), L + U)
TGS = -np.dot(np.linalg.inv(L + D), U)

print("TJ = \n", TJ)
print("TGS = \n", TGS)

rhoTJ = max(abs(np.linalg.eig(TJ)[0]))
rhoTGS = max(abs(np.linalg.eig(TGS)[0]))

print(f"\nrho(TJ) = {rhoTJ:0.4f}")
print(f"rho(TGS) = {rhoTGS:0.4f}")
```
